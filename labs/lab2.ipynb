```python
import csv
from pathlib import Path


csv_path = Path("../dataset/labels.csv").resolve()

print(csv_path)
print(csv_path.exists())
```

```python
import pandas as pd
from collections import defaultdict
df = pd.read_csv(csv_path)
count = defaultdict(int)
labels = df["label"].tolist()
for label in labels:
    count[label] += 1
print(count.values())
```
