```python
# ============================================
# 1) Imports and random seed
# ============================================

import numpy as np

# California housing dataset (regression problem)
from sklearn.datasets import fetch_california_housing

# Split dataset into training and testing sets
from sklearn.model_selection import train_test_split

# Linear Regression model
from sklearn.linear_model import LinearRegression

# Regression evaluation metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


# Set seed for reproducibility
# Ensures the same train/test split every time
SEED = 42
np.random.seed(SEED)


# ============================================
# 2) Load dataset
# ============================================

# California housing dataset:
# - Each sample represents a district in California
# - Features: population, median income, average rooms, etc.
# - Target: median house value (continuous real number)

data = fetch_california_housing()

X = data.data      # Feature matrix (n_samples × n_features)
y = data.target    # Continuous target values

print("X shape:", X.shape)
print("y shape:", y.shape)


# ============================================
# 3) Train-test split
# ============================================

# 80% training data
# 20% testing data
# random_state ensures reproducibility

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=SEED
)


# ============================================
# 4) Train Linear Regression model
# ============================================

# Linear regression assumes a linear relationship:
#
#     ŷ = w^T x + b
#
# The model learns weights w and bias b by minimizing
# the Mean Squared Error (least squares solution).

model = LinearRegression()

# Fit model on training data:
# This computes the optimal weights and bias
model.fit(X_train, y_train)


# ============================================
# 5) Evaluate model
# ============================================

# Predict house prices for unseen test data
y_pred = model.predict(X_test)

# --- Evaluation Metrics ---

# MAE (Mean Absolute Error):
# Average absolute difference between true and predicted values
mae = mean_absolute_error(y_test, y_pred)

# MSE (Mean Squared Error):
# Average squared difference (penalizes large errors more strongly)
mse = mean_squared_error(y_test, y_pred)

# RMSE (Root Mean Squared Error):
# Square root of MSE (same units as target variable)
rmse = np.sqrt(mse)

# R² (Coefficient of Determination):
# Measures proportion of variance explained by the model
# R² = 1 → perfect fit
# R² = 0 → no better than predicting the mean
r2 = r2_score(y_test, y_pred)


print("MAE:", mae)
print("RMSE:", rmse)
print("R^2:", r2)

```
