```python
# ===============================
# 1) Import required libraries
# ===============================

import numpy as np

# Dataset
from sklearn.datasets import load_breast_cancer

# Splitting data
from sklearn.model_selection import train_test_split

# Feature scaling
from sklearn.preprocessing import StandardScaler

# Logistic Regression model
from sklearn.linear_model import LogisticRegression

# Pipeline allows chaining multiple steps together
from sklearn.pipeline import Pipeline

# Evaluation metrics
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report


# ===============================
# 2) Set random seed
# ===============================

# The seed ensures reproducibility.
# Every time we run this notebook, we get the same train/test split.
SEED = 42
np.random.seed(SEED)


# ===============================
# 3) Load dataset
# ===============================

# Breast cancer dataset:
# Binary classification (malignant vs benign)
data = load_breast_cancer()
print("data type : ", type(data))
print("data.keys : " ,data.keys())
X = data.data      # Features (measurements)
y = data.target    # Labels (0 or 1)


print("Classes:", data.target_names)
print("X shape:", X.shape)


# ===============================
# 4) Split dataset into train/test
# ===============================

# 80% training, 20% testing
# stratify=y keeps class proportions balanced in both sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=SEED,
    stratify=y
)

# ===============================
# 5) Build Pipeline (IMPORTANT)
# ===============================

# A Pipeline chains multiple steps into one object.
# Here we do TWO things:
#
# Step 1: StandardScaler()
#   - Standardizes features (mean=0, std=1)
#   - Logistic regression works better when features are scaled.
#
# Step 2: LogisticRegression()
#   - Learns weights w and bias b
#   - Uses sigmoid function to model probability
#
# The pipeline ensures:
#   - Scaling is learned ONLY from training data
#   - Same scaling is applied to test data automatically
#   - Cleaner and safer workflow

clf = Pipeline([
    ("scaler", StandardScaler()),   # Step 1: scale features
    ("logreg", LogisticRegression(
        max_iter=2000,              # Increase iterations to ensure convergence
        random_state=SEED           # For reproducibility
    ))
])


# ===============================
# 6) Train the model
# ===============================

# This does TWO things internally:
# 1) Fit scaler on X_train
# 2) Transform X_train
# 3) Train logistic regression on scaled data

clf.fit(X_train, y_train)


# ===============================
# 7) Evaluate the model
# ===============================

# Predict class labels on unseen test data
y_pred = clf.predict(X_test)

# Accuracy: percentage of correct predictions
print("Accuracy:", accuracy_score(y_test, y_pred))

# Confusion matrix:
# [TN FP]
# [FN TP]
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred))

# Detailed metrics:
# Precision, Recall, F1-score for each class
print("\nReport:\n", classification_report(
    y_test,
    y_pred,
    target_names=data.target_names
))
```
